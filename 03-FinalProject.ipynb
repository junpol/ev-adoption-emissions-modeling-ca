## Background and Prior Work

California puts a huge share on cleaning up road travel, because transportation is the biggest source of greenhouse gases,about 39% of total emissions in 2022.<sup><a href="#fn1">1</a></sup> In 2022, the California Air Resources Board (CARB) passed the Advanced Clean Cars II (ACC II) rule, which sets the roadmap for California to get to 100% zero emission new car sales by 2035 (including electric, plug-in hybrid, and hydrogen vehicles).<sup><a href="#fn2">2</a></sup> Starting in 2026, at least 35% of new car sales have to be ZEVs, then 68% in 2030, and 100% by 2035.


It may sound ambitious, but it has to be. The ACC II rule is one of California’s biggest tools to actually hit its climate goals, and it comes with massive benefits: fewer smog-forming emissions, lower carbon output, and cleaner air for communities that have been hit hardest by pollution. CARB estimates the rule will cut 395 million metric tons of greenhouse gases between 2026 and 2040, avoiding 915 million barrels of oil being burned. For people living near freeways and urban areas, that’s not just a climate win, it’s a public health one.


However, right now, the policy faces uncertainty. In May 2025, the U.S. Senate voted to revoke California’s authority to set its own clean-air rules under the Clean Air Act, something that’s never happened before in the law’s 60-year history.<sup><a href="#fn3">3</a></sup> The vote is a reminder that even as technology moves forward, climate progress still depends on political support—and for California, this fight determines whether its zero-emission goals can even stay on track. For me, that’s what makes this question so relevant: it’s more important than ever to understand, with real data, how big of a difference EV adoption can actually make.


For this project, we want to use a data-driven approach to estimate how many Californians would need to switch to electric cars by 2035 to make a noticeable change in CO₂ emissions. A good benchmark is a 10% statewide reduction in CO₂, which would mean roughly a 25–30% cut in transportation emissions alone (assuming other sectors stay constant).<sup><a href="#fn1">1</a></sup> Light-duty vehicles,like passenger cars and small trucks, make up most of this category, so the real question becomes: how much can replacing gas cars with ZEVs actually move the needle?


California’s already moving in that direction. As of early 2025, about one out of every four new cars sold in the state is a zero-emission vehicle.<sup><a href="#fn4">4</a></sup><sup><a href="#fn5">5</a></sup> Using that, plus vehicle-miles-traveled data from Caltrans, you can estimate how much gas is displaced—and from there, calculate avoided CO₂.<sup><a href="#fn6">6</a></sup> The EPA gives a standard conversion: 8,887 grams of CO₂ per gallon of gasoline burned.<sup><a href="#fn7">7</a></sup> Multiply that by gallons avoided and you get tailpipe CO₂ savings. To line things up with state data, these reductions should be compared against CARB’s official GHG inventory, which tracks sector-level emissions every year.<sup><a href="#fn8">8</a></sup>


Since EVs shift emissions from tailpipes to the power grid, the electricity mix also matters. Under Senate Bill 100 (SB 100), California plans to reach 100% clean electricity by 2045, and it’s already making steady progress.<sup><a href="#fn9">9</a></sup> That means every year, electric miles will likely get “cleaner” than gas miles.


From a modeling perspective, researchers at UC Davis and the National Center for Sustainable Transportation (NCST) suggest combining ZEV adoption forecasts, fleet-turnover rates, VMT data, and grid-emission factors to build a realistic forecast of emission impacts.<sup><a href="#fn10">10</a></sup> Adding variables like charging access and driving behavior helps show what’s actually possible under different scenarios.


In short, this project explores that exact intersection: policy, behavior, and data. By forecasting EV adoption under different assumptions—like faster sales ramps or cleaner electricity, we can see how many Californians would need to go electric for the state to make a real, measurable dent in CO₂ by 2035. And given how uncertain the policy landscape just got, it feels like the perfect time to ask that question.

### Vehicle Fuel Type Count by ZIP Code (2024)

This dataset gives us a full picture of what kinds of vehicles people in California are actually driving as of the end of 2024. For every ZIP code, it tells us the fuel type, model year, vehicle make, duty (light or heavy), and how many registered vehicles fall into each category. The most important metrics for our project are the number of vehicles, the fuel type they run on, whether they are light-duty, and how old they are. All of these values are straightforward counts or labels, but together they tell a very clear story about the state's current vehicle landscape.

Fuel type matters the most for us because it directly connects to CO2 emissions. A "Gasoline" or "Diesel" vehicle means tailpipe emissions every mile it's driven, while "Electric" and some hybrids mean much lower or zero tailpipe emissions. Even though this dataset doesn't include emissions in grams or tons, fuel type acts as a strong indicator. For example, every gallon of gasoline burned emits about 8,887 grams of CO2, so knowing how many gas cars exist today is basically the starting point for estimating how much emissions could be avoided if those cars were replaced by EVs.

The duty classification is also important. Light-duty vehicles are the regular cars and SUVs most Californians drive every day, and they're responsible for most transportation emissions. Heavy-duty vehicles behave differently and contribute differently, so focusing on light-duty vehicles makes our estimates more accurate. Model year is another metric that helps us a lot. The age of a vehicle tells us something about how likely it is to retire naturally between now and 2035. Older cars tend to pollute more, but they also might be gone soon anyway, which changes how many vehicles we actually need to convert for our scenarios.

Even though the dataset is incredibly helpful, it does come with a few drawbacks. The biggest limitation is that it doesn't tell us how much each vehicle is driven. A ZIP code with a lot of cars doesn't necessarily produce more emissions if people there drive fewer miles. Since CO2 emissions depend heavily on driving behavior, we will need to combine this dataset with Caltrans VMT (Vehicle Miles Traveled) data to get a more realistic picture.

Another issue is that anything older than 2011 is grouped together as "<2011," which makes it harder to understand the age distribution of the oldest vehicles. Those older cars usually pollute the most, so losing that detail isn't ideal. The dataset also only captures one moment in time (2024). It doesn't show how EV adoption has changed in the last few years or how gas car numbers are trending. For our 2035 projections, we will need to connect this to earlier datasets to understand growth patterns instead of guessing.

Finally, the dataset doesn't include fuel economy, actual emissions, or how energy usage might change as more EVs hit the grid. We will have to pull those numbers from EPA and CARB sources and merge them ourselves.

Even with those limitations, this dataset is still one of the most important foundations for our entire project. It tells us where California is starting from. Without knowing the current mix of gas cars, EVs, and hybrids, we wouldn't be able to estimate how many people need to switch to electric vehicles to make a real dent in statewide CO2 levels by 2035.

#### **Section 1**
##### **Predicting 2035 California CO2 emissions**

To answer our research question, we will first need a prediction into the future of the C02 concentration rate in 2035. To achieve this, we will look at the California GHG Emission dataset. Plotting the CO2 emission overtime and fitting a regression model will give us said predicition.

Let us first look at the raw observations:

```python
# Load in the data
df_ghg = pd.read_csv('./data/02-processed/ca-ghg-emissions-wrangled.csv')
sns.scatterplot(data=df_ghg, x='Year', y='California Emissions')
```

As we can see, California Emissions are already on a downward trend and the shape does not seem to be quite linear. Because we want to have an accurate prediction that is generalized to the data, we need to pick a model that will best fit this data. We will use polynomial regresssion to do so.

```python
X = df_ghg['Year']
y = df_ghg['California Emissions']
coeffs = np.polyfit(X, y, deg=3)    # We use degree 3, as we feel it balances overfitting with generalization of this data
model = np.poly1d(coeffs)

# Predict y values of given time period and a decade later
X_pred = [*X, *np.linspace(2025, 2036, 10)]
y_pred = model(X_pred)       # Predict y values with this model

# Plot all data points and predictions
plt.plot(X_pred, y_pred, label='Predicted')
plt.scatter(X, y, label='Known')
plt.plot([2000, 2036], [y_pred[-2], y_pred[-2]], label='2035 Projection', linestyle=':')
plt.plot([2000, 2036], [y_pred[-2] * (1 - 0.25), y_pred[-2] * (1 - 0.25)], label='25% Reduction Goal', linestyle=':')

plt.xlabel('Year')
plt.ylabel('California CO2 Emissions (Million Metric Tons)')
plt.legend()


# Print out the predicted C02 emission for 2035
print(f'Predicted CO2 emission for 2035: {y_pred[-1]}')
```

From these results, we find the predicted CO2 emission for 2035 to be around 256.457 MMT.

For the purposes of our research question, we want statewide C02 levels to drop 25% by 2035 relative to its projected level. That is, we want a CO2 reduction of $0.25 * 256.457=64.114$ MMT down to $192.343$ MMT as a result of decreasing vehicle emissions. According to the [current California GHG Emission Inventory Data](https://ww2.arb.ca.gov/ghg-inventory-data), we can see that 38% of all Californian CO2 emissions are produced by transportation. This implies that $0.38 * 256.457 = 97.454$ MMT of CO2 are to be produced by vehicles for the year 2035. Thus, we can say that vehicle emissions need to drop by $\frac{64.114}{97.454}=65.79\%$ in order to meet our goal of reducing statewide emissions by 25%. Note that we are making the assumption that C02 levels are linearly related to the vehicle count, which may affect the validity of our results.

Before we determine the population of vehicles needed to swap to low-emission vehicles, we need to figure out how much of each car's fuel type (e.g. gas, diesel, electric, etc.) is on the road in California, as well as which of these types contribute more or less to the total CO2 emission of vehicles.

#### **Determining CO2 emission contribution based on car fuel type**

First off, we shall determine how much a given fuel type contributes to the total CO2 emission of vehicles, since diesel vehicles will tend to release more CO2 (and thus more detrimental to the environment) than gasoline or hybrid vehicles. To do this, we will look at each fuel types 'emission coefficent'. That is, the amount of CO2 released per unit of volume of its usage.

According to the [U.S. Energy Information Adminstration](https://www.eia.gov/environment/emissions/co2_vol_mass.php), motor gasoline has a CO2 emission coefficient of 9.46 kg/gallon, whereas diesel has an emission coefficient of 10.19 kg/gallon. By the nature of hybrid vehicles, we will give these cars half the gasoline emission coefficient, 4.73 kg/gallon. While hybrids may vary in their CO2 output, we make the assumption that  on average, they produce half the amount to that of gasoline vehicles, although this claim is weakly statisically supported and may perturb our results from the true result. Morever, electric vehicles have a coefficient of 0, naturally. With these values, we can assign respective cars belonging to a fuel type their appropriate emission coefficient to help breakdown their CO2 contributions. 

Now that we have each fuel type's emission factor, let us now determine the distribution of diesel, gasoline, hybrid, and electric cars on the road. To do this, we will need to look at all cars in California with the help of the [Vehicle Fuel Type Count by Zip](#vehicle-fuel-type-count-by-zip-code-2024) dataset.

```python
# Obtain vehicle-by-zip dataset
def combine_vehicle_zip():
    PROCESSED_DIR = "./data/02-processed/"
    PROCESSED_DEST = [
        PROCESSED_DIR + "vehicle-by-zip-2025.csv",
        PROCESSED_DIR + "vehicle-by-zip-2024.csv",
        PROCESSED_DIR + "vehicle-by-zip-2023.csv",
        PROCESSED_DIR + "vehicle-by-zip-2022.csv",
        PROCESSED_DIR + "vehicle-by-zip-2021.csv",
        PROCESSED_DIR + "vehicle-by-zip-2020.csv",
        PROCESSED_DIR + "vehicle-by-zip-2019.csv",
    ]
    combined = []
    for path in PROCESSED_DEST:
        combined.append(pd.read_csv(path))
    all_years = pd.concat(combined, ignore_index=True)
    return all_years

df_zip = combine_vehicle_zip()
```

Let us plot the distribution of cars across California:

```python
# Group the cars by fuel group
fuel_group_map = {'Gasoline': 'Gasoline',
                  'Hybrid Gasoline': 'Hybrid',
                  'Plug-in Hybrid': 'Hybrid',
                  'Diesel and Diesel Hybrid': 'Diesel',
                  'Battery Electric': 'Electric',
                  'Flex-Fuel': 'Other',
                  'Natural Gas': 'Other',
                  'Hydrogen Fuel Cell': 'Other',
                  'Other': 'Other',
                  'Unk': 'Other'}
fuel_group = [fuel_group_map[x] for x in df_zip['fuel']]
df_zip['fuel_group'] = fuel_group

# Plot value counts by fuel group
sns.countplot(data=df_zip, y='fuel_group', stat='proportion')
plt.xlabel('Proportion')
plt.ylabel('Fuel Type')
```

As we only have emission factors for gasoline, hybrid, electric, and diesel vehicles, we can only analyze these four types. Consequently, the remaining 'Other' cars will be lost, which will also affect our results.

```python
# Print and plot value counts by fuel group, EXCLUDING Other
print('Fuel Group Proportions\n', df_zip[df_zip['fuel_group'] != 'Other']['fuel_group'].value_counts(normalize=True))
print('Total Gasoline-Powered Vehicles:', df_zip['fuel_group'].value_counts()['Gasoline'])
sns.countplot(data=df_zip[df_zip['fuel_group'] != 'Other'], y='fuel_group', stat='proportion')
plt.xlabel('Proportion')
plt.ylabel('Fuel Type')
```

From this output, we can see that there is an overwhelming majority of gasoline cars with a total of 2,489,105 vehicles. The distribution of cars according to their fuel type is 66.3% gasoline, 14.7% diesel, 14.2% Hybrid, and 5.8% electric. Together with their emission factors, we can now calculate their overall contribution to the transportation CO2 levels. We can do this by multiplying this distribution by their respective emission coefficient, then calculating the end proportion. In particular, the emission coefficients serve as weights within the distribution.

Recall each fuel types' emission factor:

- Diesel: 10.19
- Gasoline: 9.46
- Hybrid: 4.73
- Electric: 0

Emission Coefficient $\times$ Distribution:
- Diesel:   $10.19 * 0.147 = 1.497$
- Gasoline: $9.46 * 0.663 = 6.272$
- Hybrid:   $4.73 * 0.142 = 0.672$
- Electric: $0$

Total: 1.497 + 6.272 + 0.672 = 8.441


Overall Contribution/Proportion:

- Diesel: $1.497 / 8.441 = 0.177$
- Gasoline: $6.272 / 8.441 = 0.743$
- Hybrid:  $0.672 / 8.441 = 0.08$
- Electric: $0$

It is now evident that gasoline vehicles produce $74.3\%$ of the CO2 emissions from transportation. From our previous calculations in predicting 2035 emissions, we determined that $65.79\%$ of transportation CO2 emissions needed to be cut in order to fulfill the $25\%$ reduction. Then, if we are cutting from emissions produced by gasoline-powered vehicles, these emissions need to be reduced by $\frac{65.79}{74.3}=88.5\%$. With this proportion, we can now answer our research question.

The [California Department of Motorvehicles](https://www.dmv.ca.gov/portal/file/california-dmv-statistics-pdf/) finds that there are a total of 35,983,261 registered vehicles in California as of the end of 2024. We claim that 66.3% of all Californian cars are fueled by gasoline, so we estimate there are around $0.663 * 35,983,261 = 22,777,404$ of these vehicles. Because we assume that CO2 transportation emissions are linearly related to vehicle count, the amount of gasoline-powered cars needed to satisfy the 25% CO2 reduction is $0.885 * 22,777,404 = 20,158,003$. In other words, we can say that 20,158,003 Californians who drive gasoline-powered vehicles would need to convert to electric vehicles in order to reduce the 2035 projected CO2 emission level by 25%. 

### Section 2: Predicting 2035 California Emissions using Ordinal Environmental Impact Score

We will now take a different approach to answer our research question. This time, instead of using emission coefficients, we will assign our own environmental impact score to each vehicle's fuel type.


Once we again, we load in California GHG emission dataset.

```python
# Load in the data
df_ghg = pd.read_csv('./data/02-processed/ca-ghg-emissions-wrangled.csv')
sns.scatterplot(data=df_ghg, x='Year', y='California Emissions')
```

Then, we plot the low-emission vs high-emission vehicles graph and standardize fuel-type labels. This categorizes each vehicle into either low-emission (electric, hydrogen, plug-in hybrid, PHEV, or fuel cell) or high-emission (hybrid, gasoline, diesel, or gas). This classification allows for simpler comparison across all years and ZIP codes.

```python
sns.set(style="whitegrid")

# ------------------------------------------------------------------
# Load processed dataset
# ------------------------------------------------------------------
df = combine_vehicle_zip()
df.head()

print("Columns in combined vehicle-zip dataset:")
print(df.columns.tolist())

# ------------------------------------------------------------------
# Clean fuel labels and define emission classes
# ------------------------------------------------------------------
df["fuel"] = df["fuel"].str.lower().str.strip()

low_emission_fuels = ["electric", "hydrogen", "plug-in hybrid", "phev", "fuel cell"]
high_emission_fuels = ["hybrid", "gasoline", "diesel", "gas"]

df["emission_class"] = np.where(
    df["fuel"].isin(low_emission_fuels),
    "Low-Emission",
    "High-Emission"
)

# Make emission_class categorical so it plots in a consistent order
df["emission_class"] = pd.Categorical(
    df["emission_class"],
    categories=["Low-Emission", "High-Emission"],
    ordered=False
)
```

Here, we generate variables that reflect environmental impact, categorizing 1 as low environmental impact and 6 as high environmental impact. This allows us to use a barplot for categorical data that will allow us to visualize the shift from high to low impact vehicles.

```python
# "Environmental impact" score used ONLY as an ordinal tag for plotting
# 1 = low-emission, 6 = high-emission (quasi-categorical, NOT used for KDE)
df["env_impact"] = np.where(df["emission_class"] == "Low-Emission", 1, 6)
df["env_impact"] = pd.Categorical(df["env_impact"], categories=[1, 6], ordered=True)

df.head()
```

This section groups the data by year and emission category. It then plots the change from high to low emission vehicles from 2019-2024. The time-series plot shows growth in the amount of low-emission vehicles, however it highlights how prevalant high-emission vehicles still are. This allows us to visualize a trend before projecting into future data.

```python
# ------------------------------------------------------------------
# Time series of low- vs high-emission vehicle counts
# ------------------------------------------------------------------
yearly_emissions = (
    df.groupby(["year", "emission_class"])["vehicles"]
      .sum()
      .reset_index()
)

emission_trends = yearly_emissions.pivot(
    index="year",
    columns="emission_class",
    values="vehicles"
)

plt.figure(figsize=(10, 6))
plt.plot(
    emission_trends.index,
    emission_trends["Low-Emission"],
    marker="o",
    label="Low-Emission (EV + Hydrogen + PHEV)"
)
plt.plot(
    emission_trends.index,
    emission_trends["High-Emission"],
    marker="o",
    label="High-Emission (Gasoline + Diesel + Hybrid)"
)

plt.title("Ownership of High vs Low Emission Vehicles in California (2019–2024)")
plt.xlabel("Year")
plt.ylabel("Total Registered Vehicles")
plt.grid(True, axis="y")
plt.legend()
plt.tight_layout()
plt.show()

emission_trends
```

This code block calculates the current number of high and low emission vehicles in California. It then computes how many high-emission vehicles would need to convert to low-emission to achieve a 25% reduction in emissions. This allows us to see the required amount before we model how many vehicle conversions are needed before 2035.

```python
# ------------------------------------------------------------------
# Current mix of high vs low emission vehicles
# ------------------------------------------------------------------
df["emission_group"] = np.where(df["emission_class"] == "Low-Emission", "low", "high")
df["emission_group"] = pd.Categorical(df["emission_group"], categories=["low", "high"])

current_counts = df.groupby("emission_group")["vehicles"].sum()
current_high = current_counts["high"]
current_low = current_counts["low"]
current_total = current_high + current_low

print("\nCURRENT CALIFORNIA VEHICLE MIX:")
print(f"High-emission vehicles: {current_high:,.0f} ({current_high/current_total:.1%} of total)")
print(f"Low-emission vehicles:  {current_low:,.0f} ({current_low/current_total:.1%} of total)")

# Assume a 25% reduction in emissions from high-emission vehicles
target_high = current_high * 0.75
vehicles_to_convert = current_high - target_high

print(f"\nVehicles that must convert to hit a 25% CO₂ reduction: {vehicles_to_convert:,.0f}")
```

To project our future scenario, we distribute the required conversions across all high-emission ZIP-fuel groups proportional to their current size. The same amount of vehicles is then added to the low-emission category. This creates a 2035 vehicle makeup under the assumption of a 25% emissions reduction from high-emission vehicles.

```python
# ------------------------------------------------------------------
# Build projected 2035 fleet under 25% reduction assumption
# ------------------------------------------------------------------
future_df = df.copy()
future_df["year"] = 2035  # tag all as 2035 scenario

# Each high-emission observation gets a share of the total high-emission fleet
future_df["high_emission_share"] = np.where(
    future_df["emission_group"] == "high",
    future_df["vehicles"] / current_high,
    0
)

# Convert that same fraction of vehicles within each high-emission cell
future_df["future_conversion"] = future_df["high_emission_share"] * vehicles_to_convert

future_df["vehicles_2035"] = future_df["vehicles"]
future_df.loc[future_df["emission_group"] == "high", "vehicles_2035"] -= future_df.loc[
    future_df["emission_group"] == "high", "future_conversion"
]
future_df.loc[future_df["emission_group"] == "low", "vehicles_2035"] += future_df.loc[
    future_df["emission_group"] == "low", "future_conversion"
]

# Make sure there are no negative vehicle counts after conversion
future_df["vehicles_2035"] = future_df["vehicles_2035"].clip(lower=0)

future_totals_check = future_df.groupby("emission_group")["vehicles_2035"].sum()
print("\nPROJECTED 2035 MIX (after 25% reduction from high-emission):")
print(f"High-emission vehicles: {future_totals_check['high']:,.0f}")
print(f"Low-emission vehicles:  {future_totals_check['low']:,.0f}")
```

This first barplot compares the current number of high and low emission vehicles with the projected 2035 numbers.

```python
# ------------------------------------------------------------------
# FIGURE 1: Current vs projected vehicle counts by emission group
# ------------------------------------------------------------------
current_totals = (
    df.groupby("emission_group")["vehicles"]
      .sum()
      .reset_index()
      .rename(columns={"vehicles": "count"})
)
current_totals["scenario"] = "Current"

future_totals = (
    future_df.groupby("emission_group")["vehicles_2035"]
      .sum()
      .reset_index()
      .rename(columns={"vehicles_2035": "count"})
)
future_totals["scenario"] = "Projected 2035 (25% reduction)"

comparison = pd.concat([current_totals, future_totals], ignore_index=True)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=comparison,
    x="emission_group",
    y="count",
    hue="scenario"
)
plt.title("Current vs Projected 2035 Vehicle Counts by Emission Group")
plt.xlabel("Emission Group (low = low-emission, high = high-emission)")
plt.ylabel("Number of Vehicles")
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.legend()
plt.show()
```

Using the environmental impact score (1 = low, 6 = high), this section compares the number of vehicles in each category today versus our 2035 projetion. The barplot shows a shift away from high-impact vehicles under this assumed reduction scenario.

```python
# ------------------------------------------------------------------
# FIGURE 2: Current vs projected environmental impact distribution
# using barplots
# ------------------------------------------------------------------
current_env = (
    df.groupby("env_impact")["vehicles"]
      .sum()
      .reset_index()
      .rename(columns={"vehicles": "count"})
)
current_env["scenario"] = "Current"

future_env = (
    future_df.groupby("env_impact")["vehicles_2035"]
      .sum()
      .reset_index()
      .rename(columns={"vehicles_2035": "count"})
)
future_env["scenario"] = "Projected 2035 (25% reduction)"

env_comparison = pd.concat([current_env, future_env], ignore_index=True)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=env_comparison,
    x="env_impact",
    y="count",
    hue="scenario"
)
plt.title("Current vs Projected Environmental Impact (1 = Low, 6 = High)")
plt.xlabel("Environmental Impact Score")
plt.ylabel("Number of Vehicles")
plt.ticklabel_format(axis="y", style="plain")
plt.tight_layout()
plt.legend(title="Scenario")
plt.show()
```
